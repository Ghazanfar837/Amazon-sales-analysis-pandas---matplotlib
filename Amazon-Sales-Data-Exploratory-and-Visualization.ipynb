### **Date Time Analyses**

### **Convert OrderDate to DateTime**

```
df["OrderDate"]=pd.to_datetime(df["OrderDate"])
```

### **Extract Year**

```
df["Year"]=df["OrderDate"].dt.year
df.head()
```

### **Extract Month Name**

```
df["Month"]=df["OrderDate"].dt.month_name()
df.head()
```

### **Extract Day Name**

```
df["Day"]=df["OrderDate"].dt.day_name()
df.head()
```

### **Order Between Two Dates**

```
df[(df["OrderDate"]>="2024-01-01")& (df["OrderDate"]<="2024-06-30")]
```

### **Monthly Sales Analyses**

## **Data Visualization using Matplotlib**

### **Line Chart-Monthly Sales Trend**

```
monthly_sales=df.groupby("Month")["TotalAmount"].sum()
plt.figure(figsize=(10,5))
plt.plot(monthly_sales.index,monthly_sales.values,marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.grid(True)
plt.show()
```

### **Bar Chart- Sales by Category**

```
category_sales=df.groupby("Category")["TotalAmount"].sum()
plt.figure(figsize=(8,5))
plt.bar(category_sales.index,category_sales.values)
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

### **Horizontal Bar Chart-Top 10 Brands**

```
top_brands=df.groupby("Brand")["TotalAmount"].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(8,6))
plt.barh(top_brands.index,top_brands.values)
plt.title("Top 10 Brands by Sales")
plt.xlabel("Total Sales")
plt.show()
```

### **Pie Chart-Sales Share by Category**

```
category_sales=df.groupby("Category")["TotalAmount"].sum()
plt.figure(figsize=(7,7))
plt.pie(category_sales.values,labels=category_sales.index,autopct="%1.1f%%",startangle=90)
plt.title("Sales Share by Category")
plt.show()
```

### **Histogram- Distribution of Order Amount**

```
plt.figure(figsize=(8,5))
plt.hist(df["TotalAmount"],bins=20)
plt.title("Distribution of Order Amount")
plt.xlabel("Total Amount")
plt.ylabel("Frequency")
plt.show()
```

### **Scatter Plot-Quantity vs Total Amount**

```
plt.figure(figsize=(8,5))
plt.scatter(df["Quantity"],df["TotalAmount"])
plt.title("Quantity vs Total Amount")
plt.xlabel("Quantity")
plt.ylabel("Total Amount")
plt.show()
```
